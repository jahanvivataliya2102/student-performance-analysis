```python
#identify which numerical features have the strongest correlation with final grades (G3).

correlation = df.corr(numeric_only=True)

g3_correlation = correlation["G3"].sort_values(
    ascending=False
)

print(g3_correlation)
```

```python
#Visualizing the correlation with G3

g3_corr = df.corr(numeric_only=True)["G3"].drop("G3")

g3_corr = g3_corr.sort_values()

plt.figure(figsize=(10, 8))

sns.barplot(
    x=g3_corr.values,
    y=g3_corr.index
)

plt.title("Correlation of Features with Final Grade (G3)")
plt.xlabel("Correlation Coefficient")
plt.ylabel("Features")

plt.axvline(x=0, color="black")

plt.show()
```

```python
#Analyzing school attendance

def attendance_category(absences):
    if absences <= 5:
        return "Low"
    elif absences <= 15:
        return "Moderate"
    else:
        return "High"

df["attendance"] = df["absences"].apply(attendance_category)

print(df["attendance"].value_counts())
```

```python
#comparing finbal grades

attendance_analysis = df.groupby("attendance")["G3"].agg(
    ["count", "mean", "median", "min", "max"]
)

print(attendance_analysis)
```

```python
# visulizing the results 

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="attendance",
    y="G3",
    order=["Low", "Moderate", "High"]
)

plt.title("Attendance vs Final Grades")
plt.xlabel("Attendance Category")
plt.ylabel("Final Grade (G3)")

plt.show()
```

### Finding 6: Attendance and Student Performance

Students with moderate absences have the highest average final grade of 11.07, followed by students with low absences (10.17) and high absences (10.03).

The results do not show a consistent relationship between attendance and final grades. Other factors may also contribute to differences in academic performance.

```python
# Pass/Fail analysis across different features
# Internet access vs result

internet_result = pd.crosstab(
    df["internet"],
    df["result"],
    normalize="index"
) * 100

print(internet_result)
```

```python
# visulization 

internet_result.plot(
    kind="bar",
    stacked=True,
    figsize=(8, 5)
)

plt.title("Pass and Fail Percentage by Internet Access")
plt.xlabel("Internet Access")
plt.ylabel("Percentage")
plt.legend(title="Result")

plt.xticks(rotation=0)
plt.show()
```

```python
# school vs result

school_result = pd.crosstab(
    df["school"],
    df["result"],
    normalize="index"
) * 100

print(school_result)
```

```python
# attendance vs result

attendance_result = pd.crosstab(
    df["attendance"],
    df["result"],
    normalize="index"
) * 100

print(attendance_result)
```

### Finding 10: Pass Percentage Across Student Groups

- Internet Access: Students with internet access have a pass percentage of 68.39%, compared with 60.61% for students without internet access.

- School: GP has a pass percentage of 67.62%, while MS has 63.04%.

- Attendance: Students with high absences have the lowest pass percentage of 54.55%. Students with low and moderate absences have similar pass percentages of approximately 68%.

These results show some differences in academic performance across student groups. However, they do not establish that these factors directly cause students to pass or fail.

```python
# Pass/Fail percentage by number of previous failures
failure_result = pd.crosstab(
    df["failures"],
    df["result"],
    normalize="index"
) * 100

print(failure_result)
```

```python
# visulization 
failure_result.plot(
    kind="bar",
    stacked=True,
    figsize=(8, 5)
)

plt.title("Pass and Fail Percentage by Previous Failures")
plt.xlabel("Number of Previous Failures")
plt.ylabel("Percentage")
plt.legend(title="Result")
plt.xticks(rotation=0)
plt.show()
```

### Pass and Fail Percentage by Previous Failures

Students with no previous failures have the highest pass percentage (75%).

As the number of previous failures increases, the pass percentage generally decreases. Students with 2 previous failures have the lowest pass percentage (17.65%).

This suggests that previous academic failures are associated with lower final performance in this dataset.

## Final Conclusion

After analyzing the student performance dataset, I found a few interesting patterns:

1. G2 (second-period grade) has the strongest relationship with the final grade (G3).
2. G1 (first-period grade) also has a strong positive relationship with G3.
3. Students with more previous academic failures generally have lower pass percentages.
4. Students with internet access have a slightly higher average final grade in this dataset.
5. The pass percentage is slightly higher among students who plan to pursue higher education.
6. Attendance categories and other factors show some differences, but not always a clear pattern.

Overall, previous academic performance appears to be strongly related to final mathematics grades. Other factors also show some differences, but this analysis does not prove that they cause changes in student performance.

## Limitations

1. This analysis uses only the student mathematics dataset, which contains 395 records.
2. The results may not represent students from other schools or regions.
3. Correlation does not always mean that one factor causes another.
4. Some categories contain fewer students than others, which can affect comparisons.
5. The pass/fail classification uses 10 as the threshold for this project.
6. The attendance categories are based on the number of absences, not directly recorded attendance percentages.
